```python
# list comprehension
list_1 = [1,2,3,4,5,6,7]
squared = [x**2 for x in list_1]
squared
```

```python
list_1 = [1,2,3,4,5,6,7]
list_2 = ['a','b','c','d','e','f','g']
pairs = [{i,j} for i in list_1 for j in list_2]
pairs
```
