## Sets

```python
# Sets in Python
# A Set is basically a set with non repeating items
```

```python
myset = {1,2,3,4,5,6,7}
myset
```

```python
newset = set([1,2,3,34,45,6,7,7,8])
newset
```

### Basic Set Operations

```python
myset.add(10)
myset
```

```python
myset.remove(3) # removes the item and gives error if item not present 
myset
```

```python
# Discard - removes the element if its presetn and does ot return any error if its not present 
myset.discard(11) # did nothing 
myset
```

```python
removed_element = myset.pop() # first in First Out 
removed_element
```

```python
myset.clear()
myset
```

### SET operations

```python
set2 = {11,12,3,4,5,6,7,23,53,65}
set2
```

```python
newset
```

```python
u_set = newset.union(set2)
u_set
```

```python
intersection = newset.intersection(set2)
intersection
```

```python
newset.difference(set2) # remove common elecments
```

```python
newset.symmetric_difference(set2) # union - commmon element
```

### Methods in Sets

```python
newset.issubset(set2) # check if newset is subset of set2
```

```python
newset.issuperset(set2)
```
